```python
import polars as pl

sales = pl.read_csv('sales_data.csv')
```

```python
# 1 Which salesperson (by ID) sold the most units in total?

sales_person = (
    sales
    .group_by('Salesperson_ID')
    .agg(
        total_sold = pl.sum('Units_Sold')
    )
    .sort(pl.col('total_sold')).reverse()
)

sales_person
# 61
```

```python
# 2 Which product (by name) was sold the most in terms of total units? (note: your answer should be  in the form Product_Number, ie Product_1051)
products = (
    sales
    .group_by('Product_Name')
    .agg(
        total_units = pl.sum('Units_Sold')
    )
    .sort(pl.col('total_units')).reverse()
)

products
# Product_1761
```

```python
# 3 Which region had the highest dollar sales in the month of January?

sales1 = sales.filter(pl.col('Month') == 1)

sales_jan = (
    sales1
    .group_by('Region')
    .agg(
        total_units = pl.sum('Total_Sale')
    )
    .sort(pl.col('total_units')).reverse()
)

sales_jan
# east
```

```python
# 4 Which salesperson (by ID) sold the fewest units in the North region?

sales_north = sales.filter(pl.col('Region') == 'North')

sales_person_north = (
    sales_north
    .group_by('Salesperson_ID')
    .agg(
        total_units = pl.sum('Units_Sold')
    )
    .sort(pl.col('total_units'))
)

sales_person_north
# 63
```

```python
# 5 What is the average unit price for products sold by Sales person 14?
# Note: For this question, assume that the "average unit price" refers to the simple average price of all products sold by Salesperson 14. 

sales_14 = sales.filter(pl.col('Salesperson_ID') == 14)

sales_14_price = sales_14["Unit_Price"].mean()

sales_14_price
# 261
```

```python
# 6 Someone tells you: the simple price averages computed above are not accurate representatives of the "average" price in a market. 
# For example, say there are only two products 1 and 2 with unit prices $20 and $500. 
# Suppose you sell 50 units of the first product and only 5 units of the second product. 
# Then (20+500)/2 = $260 is the simple price average. 
# That is clearly not representative of the typical price paid in the market since 50/55 = 91%  of all products sold were priced at $20! 
# A better average is a sales-weighted average, ie (20*50 + 5*500)/(50 + 5) = $63.6. This average gives higher weight to products that had higher sales.
# Calculate the average sales-weighted unit price for products sold by Sales person 14 (round to nearest integer)

# Calculate the sales-weighted average unit price
sales_weighted_avg_price = (
    (sales_14['Unit_Price'] * sales_14['Units_Sold']).sum() /
    sales_14['Units_Sold'].sum()
)

print(sales_weighted_avg_price)
# 266
```

```python
# 11 Finally, let's look at the effect of salesperson experience on dollar sales (Total_Sale). 
# Define "inexperienced" salespeople as having 5 years or less of sales experience and let the rest (more than 5 years of experience) be "experienced" salespeople. 
# What is the average increase in sales across all products due to experience? 
# To answer this, find the average increase in sales for experienced salespeople compared to inexperienced salespeople for each product and then compute the average increase across all products. 
# So you need to find (1) average dollar sales for each product for experienced and inexperienced salespeople (2) find the difference between the two and then (3) find the average difference across all products. 
# Report your answer as rounded to the nearest integer. 

# Step 1: Categorize salespeople by experience

sales = sales.with_columns(
    pl.when(pl.col('Salesperson_Experience') > 5)
    .then(pl.lit('Experienced'))
    .otherwise(pl.lit('Inexperienced'))
    .alias('Experience_Category')
)


print(sales.select(['Salesperson_Experience', 'Experience_Category']).head())

print(type(sales))

avg_sales_by_experience = (
    sales.
    group_by(['Product_ID', 'Experience_Category'])
    .agg([pl.col('Total_Sale').mean().alias('Avg_Total_Sale')])
    .pivot(values='Avg_Total_Sale', index='Product_ID', columns='Experience_Category')
)


# Step 3: Calculate the difference in average sales between experienced and inexperienced salespeople
avg_sales_by_experience = (
    avg_sales_by_experience
    .with_columns(
        (pl.col('Experienced') - pl.col('Inexperienced')).alias('Sales_Difference')
        )
)
print(avg_sales_by_experience.head())

# Step 4: Compute the average increase across all products
# average_increase = avg_sales_by_experience.select(pl.col('Sales_Difference').mean())[0, 0]
average_increase = avg_sales_by_experience["Sales_Difference"].mean()

print(average_increase)

# 4603 correct
```
